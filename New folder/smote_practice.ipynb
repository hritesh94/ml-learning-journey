```python
from sklearn.datasets import make_classification
```

```python
X,y=make_classification(n_samples=1000,n_redundant=0,n_features=2,n_clusters_per_class=1,
                        weights=[0.90],random_state=12)
#Create 1000 artificial data points with 2 features, split them into two classes with a 90/10 imbalance,
#make one cluster for each class, don't add redundant features, and make the random generation reproducible.
#
```

```python
import pandas as pd
df1 = pd.DataFrame(X,columns = ['f1','f2'])
df2 = pd.DataFrame(y,columns = ['target'])
final_df = pd.concat([df1,df2],axis=1)
final_df.head()
```

```python
final_df['target'].value_counts()
```

```python
import matplotlib.pyplot as plt
plt.scatter(final_df['f1'],final_df['f2'],c=final_df['target'])
# c means "Color each point according to its target value, basically for target 1 use one color and for target 2 use another value"
```
