### Assignment 13: Default Dictionary

Create a default dictionary where each key has a default value of an empty list. Add some elements to the lists and print the dictionary.

```python
d = {i:[] for i in range(1,6)}
print(d)
d[1].extend([1,2,3,4])
d[2].append(3)
d[3].append(4)
d[4].append(6)
d[5].append(8)

print(d)
##above is my solution to the question

## below is the tutors solution
from collections import defaultdict
#A dictionary that automatically creates a default value when a missing key is accessed.
default_dict = defaultdict(list)
default_dict['a'].append(1)
default_dict['a'].append(2)
default_dict['b'].append(3)
print(default_dict)

# ===================== DEFAULTDICT NOTES =====================
# defaultdict is a special type of dictionary from the collections module.
#
# Syntax:
# from collections import defaultdict
# d = defaultdict(list)
#
# How it works:
# - If you access a key that doesn't exist, defaultdict automatically
#   creates it using the default value you specify.
#
# Example:
# d = defaultdict(list)
# d["A"].append(10)
#
# Behind the scenes, Python does:
# if "A" not in d:
#     d["A"] = []
# d["A"].append(10)
#
# Common default factories:
# defaultdict(list) -> []   (empty list)
# defaultdict(int)  -> 0    (integer zero)
# defaultdict(set)  -> set() (empty set)
# defaultdict(str)  -> ""   (empty string)
#
# append(x)  -> Adds ONE item to the list.
# extend(lst)-> Adds ALL items from another list.
#
# Use defaultdict when you don't want to manually initialize keys
# before adding values to them.
# ============================================================
```

### Assignment 14: Counting with Dictionaries

Write a function that takes a string and returns a dictionary with the count of each character in the string. Print the dictionary.

```python
s = "hello"
def count_char(s):
    count_dict={}
    for char in s:
        count_dict[char] = count_dict.get(char,0) + 1
    return count_dict

count_char(s)
```
